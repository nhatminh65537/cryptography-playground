```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from share.interactive import Simulation
from share.gkr import *

from random import seed
```

```python
c = ArithCircuit([
    [add(0, 3), mul(1, 2)],
    [mul(0, 1), add(2, 3), mul(4, 5), add(6, 7)],
    [add(0, 1), mul(2, 3), add(4, 5), mul(6, 7), add(0, 7), mul(1, 6), add(2, 5), mul(3, 4)],
    [var(i) for i in range(8)],
])
x = [3, 1, 4, 1, 5, 9, 2, 6]

prover = GKRProver("Prover", c)
verifier = GKRVerifier("Verifier", c, x)
prover.dst = verifier.name
verifier.dst = prover.name

sim = Simulation([prover, verifier])
sim.run()
print(sim.result)
sim.print_times()
sim.print_log()
```
